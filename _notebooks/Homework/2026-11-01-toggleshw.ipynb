```python
%%html
<!-- UI_RUNNER: Homework answer - Toggle Study Desk | show_code: true -->

<section class="ocs__container" id="hw-desk">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Toggle Study Desk</h2>
    <p class="ocs__description">Three toggles that each change something on the page.</p>

    <p id="hw-count" role="status">0 of 3 toggles on</p>

    <div class="ocs__stack">
      <label class="ocs__toggle ocs__toggle--green">
        <input id="hw-focus" class="ocs__toggle-input" type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Focus mode: hide distractions</span>
      </label>

      <label class="ocs__toggle ocs__toggle--green">
        <input id="hw-hint" class="ocs__toggle-input" type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Show hint</span>
      </label>

      <label class="ocs__toggle ocs__toggle--green">
        <input id="hw-schedule" class="ocs__toggle-input" type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Show study schedule</span>
      </label>
    </div>

    <p id="hw-distractions">Distractions: phone, open tabs, snacks.</p>
    <p id="hw-hint-text" hidden>Hint: use <code>toggle.checked</code> to test if a toggle is on.</p>
    <p id="hw-schedule-text" hidden>Schedule: read 20 minutes, code 20 minutes, review 10 minutes.</p>
  </div>
</section>

<script>
const focus = document.getElementById("hw-focus");
const hint = document.getElementById("hw-hint");
const schedule = document.getElementById("hw-schedule");
const count = document.getElementById("hw-count");
const toggles = [focus, hint, schedule];

// One function reads all three toggles and updates the page.
function updatePage() {
  // Focus mode hides the distractions.
  document.getElementById("hw-distractions").hidden = focus.checked;

  // The hint and schedule appear when their toggle is on.
  document.getElementById("hw-hint-text").hidden = !hint.checked;
  document.getElementById("hw-schedule-text").hidden = !schedule.checked;

  // Count the toggles that are on.
  let on = 0;

  toggles.forEach(function (toggle) {
    if (toggle.checked) {
      on = on + 1;
    }
  });

  count.textContent = on + " of 3 toggles on";
}

// Every toggle runs the same update.
toggles.forEach(function (toggle) {
  toggle.addEventListener("change", updatePage);
});

updatePage();
</script>
```
