```python
# CODE_RUNNER: HOMEWORK
# Recovery Center Cost Report
cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4
enrollment_fee = 75

weekly_cost = cost_per_session * sessions_per_week
monthly_cost = weekly_cost * weeks_per_month
total_cost_with_fee = monthly_cost + enrollment_fee
average_daily_cost = monthly_cost / 30

print("Weekly Cost: " + str(weekly_cost))
print("Monthly Cost: " + str(monthly_cost))
print("Total Cost with Fee: " + str(total_cost_with_fee))
print("Average Daily Cost: " + str(average_daily_cost))
```
